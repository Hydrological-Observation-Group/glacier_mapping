### Dataset normalization.  
- Convert DN value to reflectance     

```python
import os 
import numpy as np
import rasterio as rio 
from glob import glob
from pyrsimg import imsShow 
import matplotlib.pyplot as plt
```

```python
### scale and offset given by GEE platform.  
scale = {'l5': 2.75e-05, 'l7': 2.75e-05, 'l8': 2.75e-05, 'l9': 2.75e-05, 's2': 0.0001} 
offset = {'l5': -0.2, 'l7': -0.2, 'l8': -0.2, 'l9': -0.2, 's2': 0} 
### max and min values for dem data
dem_max, dem_min = 8848.0, -420.0  # highest and lowest elevation on Earth: Mount Everest/Dead Sea Shore 

```

```python
### get the secene and truth pairwise data
paths_scene = sorted(glob('data/dset/scene/scene_ori/*/*.tif'))
paths_dem = sorted(glob('data/dset/dem/dem_ori/*.tif'))
paths_truth = sorted(glob('data/dset/truth/truth_tif/*/*.tif'))
print(len(paths_scene))
print(len(paths_dem)) 
print(len(paths_truth)) 
```

```python
def dn_to_ref(image, scale=0.0001, offset=0):
    '''
    des: convert DN to reflectance
    args:  numpy array
    '''
    image_nor = image.astype(np.float32) * scale + offset
    image_nor = np.clip(image_nor, 0.0, 1.0)
    return image_nor

def normalize_dem(dem_img, dem_max, dem_min):
    '''
    des: normalize dem to [0, 1]
    '''
    if dem_max <= dem_min:
        raise ValueError('max_dem must be greater than min_dem')
    dem_normalized = (dem_img.astype(np.float32) - dem_min) \
                        / (dem_max - dem_min)
    return np.clip(dem_normalized, 0.0, 1.0)

```

```python
## (1) normalization for satellite image
for path_scene in paths_scene:
    path_save = path_scene.replace('scene_ori/', 'scene_nor/')
    scene_sat = path_scene.split('/')[-1][0:2].lower()
    scene_rio = rio.open(path_scene)
    profile = scene_rio.profile.copy()
    profile.update(dtype='float32')
    scene_nor = dn_to_ref(image=scene_rio.read(), 
                       scale= scale[scene_sat], 
                       offset= offset[scene_sat])
    # write to disk   
    if os.path.exists(path_save): os.remove(path_save)
    with rio.open(path_save, 'w', **profile) as dst:
        dst.write(scene_nor.astype('float32'))
    print(f'Wrote normalized scene to: {path_save}')
    break  
```

```python
## (1) normalization for dem
for path_dem in paths_dem:
    path_save = path_dem.replace('dem_ori', 'dem_nor')
    dem_rio = rio.open(path_dem)
    profile = dem_rio.profile.copy()
    profile.update(dtype='float32') 
    dem_nor = normalize_dem(dem_img=dem_rio.read(), 
                                dem_max=dem_max, 
                                dem_min=dem_min)
    # write to disk
    if os.path.exists(path_save): os.remove(path_save)
    with rio.open(path_save, 'w', **profile) as dst:
        dst.write(dem_nor.astype('float32'))
    print(f'Wrote normalized dem to: {path_save}')
    break 

```

```python
## check scene normalization
id = 9
path_scene = paths_scene[id]   
print(path_scene)
path_scene_nor = path_scene.replace('scene_ori/', 'scene_nor/')
scene_rio, scene_nor_rio = rio.open(path_scene), rio.open(path_scene_nor)
fig, ax = plt.subplots(1, 2, figsize=(8, 3));
ax[0].hist(scene_rio.read().flatten(), bins=100, color='blue');
ax[0].set_title('original scene');  
ax[1].hist(scene_nor_rio.read().flatten(), bins=100, color='orange');
ax[1].set_title('normalized scene');
plt.tight_layout() 
```

```python
imsShow([scene_rio.read().transpose(1, 2, 0), 
                scene_nor_rio.read().transpose(1, 2, 0)],
                img_name_list=['before nor', 'after nor'], 
                clip_list=[1, 1], 
                color_bands_list=[[3, 2, 1], [3, 2, 1]],
                figsize=(7, 3));
```
